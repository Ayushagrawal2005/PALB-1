```python
import sqlite3
```

```python
###Connect to db
connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
## Create a table
cursor.execute('''
Create table if not exists employees(
                id Integer Primary Key,
                name Text Not Null,
                age Integer,
                department text
                 )
''')

connection.commit()
```

```python
### Inseet data
cursor.execute('''
Insert Into employees(name,age,department)
    VALUES ('Bob',25,'Engineering') 

''')

cursor.execute('''
Insert Into employees(name,age,department)
    VALUES ('Krish',28,'Data Science') 

''')
cursor.execute('''
Insert Into employees(name,age,department)
    VALUES ('Ayush',26,'CEO') 

''')
cursor.execute('''
Insert Into employees(name,age,department)
    VALUES ('Gopal',24,'President') 

''')
cursor.execute('''
Insert Into employees(name,age,department)
    VALUES ('Vansh',35,'Mechanichal') 

''')
connection.commit()
```

```python
## Querying the data
cursor.execute('Select * from employees')
rows = cursor.fetchall()
for row in rows:
    print(row)

```
