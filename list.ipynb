```python
#numpy vectorwise operation
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([4,5,6,7,8,])

#element wise addition
print(arr1 + arr2)

#Element wise subtraction
print(arr1 - arr2)
```

```python
### Stastical concepts normalisation
mean = np.mean(arr)
std = np.std(arr)
normalised_arr = (arr - mean) / std
print(normalised_arr)
```
