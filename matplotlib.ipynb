```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [1,4,6,8,0]
plt.plot(x,y,color='red',linestyle=':',marker='*' )
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title("Basic Line Plot")
```

```python
x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,6,8]
plt.figure(figsize=(9,5))
plt.subplot(1,2,1)
plt.plot(x,y1,color='green')
plt.title("plot 1")
```

```python
### Bar Plot
categories = ['A','B','C','D','E','F']
values = [5,7,3,4,9,2]
plt.bar(categories,values,color='green')
```

```python
Labels=['A','B','C','D']
sizes=[30,20,40,10]
colors=['gold','yellowgreen','lightcoral','skyblue']
explode = (0.2,0,0,0)
plt.pie(sizes,labels=Labels,colors=colors,autopct="%1.1f%%")
```

```python
## Sales Data Visualisation
import pandas as pd
sales_data_df = pd.read_csv('sales.csv')
sales_data_df.head(5)
```

```python
sales_data_df.info()
```

```python
total_sales_by_item = sales_data_df.groupby('Item Type')['Unit Price'].sum()
print(total_sales_by_item)
```

```python
total_sales_by_item.plot(kind='bar',color='green')
```

```python
sales_trend = sales_data_df.groupby('Order Date')['Total Revenue'].sum().reset_index()
plt.plot(sales_trend['Order Date'],sales_trend['Total Revenue'])
```

```python
sales_trend.plot(kind='bar',color='teal')
```
