```python
import pandas as pd
```

```python
data = [1, 2, 3, 4, 5]
series = pd.Series(data)
print(series)
```

```python
 # creating the series from dictonaries
data = {'a': 1, 'b': 2, 'c': 3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
# Dataframes
data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'City': ['New York', 'London', 'Tokyo']}
df = pd.DataFrame(data)
print(df)
```

```python
df = pd.read_csv('sales.csv')
df.head(5)
```

```python
# Data manipulation
salaries = [10000, 20000, 30000, 40000]
df['Salary'] = salaries * (len(df) // len(salaries)) + salaries[:len(df) % len(salaries)]
df
```
